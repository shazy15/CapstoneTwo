```python
#import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#to display all coulmns, so nothing gets hidden
pd.set_option('display.max_columns', None)
```

```python
#loading the dataset
df = pd.read_csv('../data/raw/MARVEL vs. DC - IMDB & ROTTEN TOMATOES.csv', encoding='latin1')

#to check the size of the dataset
print(f"Dataset Shape: {df.shape}")

#taking a look at the first 5 rows
df.head()
```

```python
#checking data types
df.info()
```

```python
#summary statistics
df.describe()
```

```python
#checking the number of unique values per column
df.nunique()
```

```python
#checking to see how many zeros exist in the numerical columns
zero_counts = (df[['runtime', 'imdb_rating', 'imdb_votes', 'imdb_gross', 'tomato_meter', 'tom_aud_score']] ==0 ).sum()
print("Count zeros per coulmn:")
print(zero_counts)
```

```python
#inspecting the movies where imdb_gross is 0
print("Movies with 0 gross:")
display(df.loc[df['imdb_gross'] == 0, ['title', 'year', 'entity', 'imdb_gross']])

#inspecting the movies where tomato_meter is 0
print("\nMovies with 0 tomato_meter:")
display(df.loc[df['tomato_meter'] == 0, ['title', 'year', 'entity', 'tomato_meter']])
```

```python
#replace 0s with NaN for columns where 0 means missing values
cols_to_fix = ['imdb_gross', 'tomato_meter']
df[cols_to_fix] = df[cols_to_fix].replace(0, np.nan)
```

```python
#check for duplicate rows
print(f"Number of duplicate rows: {df.duplicated().sum()}")
```

```python
#drop redunt coulmns if they exist
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print(df.columns.tolist())
```

```python
#saved cleaned up data to csv
df.to_csv('../data/processed/cleaned_marvel_dc_movies.csv', index=False)

print("Data wrangling complete! Cleaned dataset successfully saved.")
```

```python
# Aggregate key metrics to compare Marvel vs. DC
summary_table = df.groupby('entity').agg(
    Total_Movies=('title', 'count'),
    Mean_IMDb_Rating=('imdb_rating', lambda x: round(x.mean(), 2)),
    Median_IMDb_Rating=('imdb_rating', 'median'),
    Mean_Tomatoes=('tomato_meter', lambda x: round(x.replace(0, float('nan')).mean(), 1)),
    Mean_Gross_Millions=('imdb_gross', lambda x: round(x.replace(0, float('nan')).mean() / 1e6, 2))
).reset_index()

# Display the final summary table
print("=== MARVEL VS. DC: FINAL COMPARISON SUMMARY ===")
display(summary_table)
```
